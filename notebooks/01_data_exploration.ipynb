This notebook explores the resume corpus and job postings dataset. Reproduces paper Figures 2 and 3.

```python
import sys
from pathlib import Path

# Add repo root to path
sys.path.insert(0, str(Path.cwd().parent))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from core.constants import RESUME_CORPUS_PATH, JOB_POSTINGS_PATH

sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 6)
```

## 1. Load Dataset 1: `clean_resume_data.csv`

```python
df = pd.read_csv(RESUME_CORPUS_PATH)
df['Feature'] = df['Feature'].fillna('').astype(str)

print(f'Total resumes: {len(df):,}')
print(f'Unique categories: {df["Category"].nunique()}')
print(f'\nColumns: {df.columns.tolist()}')
df.head()
```

## 2. Category Distribution (paper Fig. 2)

```python
counts = df['Category'].value_counts()
print(counts)

plt.figure(figsize=(14, 6))
counts.plot(kind='bar', color='#6c63ff', edgecolor='#4a41c9')
plt.axhline(counts.mean(), color='#ff4d6d', linestyle='--', label=f'Mean = {counts.mean():.0f}')
plt.title('Resume Count per Category (Dataset 1)')
plt.xlabel('Category')
plt.ylabel('Number of Resumes')
plt.legend()
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## 3. Resume Length Statistics

```python
df['word_count'] = df['Feature'].str.split().str.len()

print(f'Mean length:   {df["word_count"].mean():.0f} words')
print(f'Median length: {df["word_count"].median():.0f} words')
print(f'Range:         {df["word_count"].min()} - {df["word_count"].max()} words')

plt.figure(figsize=(12, 5))
sns.histplot(df['word_count'], bins=50, color='#00d4aa', edgecolor='#00a082')
plt.title('Resume Word Count Distribution')
plt.xlabel('Word Count')
plt.ylabel('Frequency')
plt.tight_layout()
plt.show()
```

## 4. Load Dataset 2: `jobs_dataset_with_features.csv`

The full dataset is approximately 1M+ rows and 600 MB. We load a 50k sample for exploration.

```python
job_df = pd.read_csv(JOB_POSTINGS_PATH, nrows=50000)
print(f'Loaded {len(job_df):,} job postings (50k sample)')
print(f'Unique roles: {job_df["Role"].nunique()}')
job_df.head()
```

## 5. Top-20 Job Roles (paper Fig. 3)

```python
top20 = job_df['Role'].value_counts().head(20)

plt.figure(figsize=(12, 8))
top20.plot(kind='barh', color='#4fc3f7', edgecolor='#0288d1')
plt.title('Top-20 Job Roles in Dataset 2 (50k sample)')
plt.xlabel('Number of Postings')
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```
